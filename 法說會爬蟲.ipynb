```python
import requests
from playwright.async_api import async_playwright

IR_URL = "https://www.cathayholdings.com/holdings/ir/financial_information/results_presentation"

DATA = {
    "first_name": "",
    "last_name": "",
    "email": "",
    "telephone": ""
}

async def main():
    async with async_playwright() as p:

        browser = await p.chromium.launch(
            headless=True,
            args=["--no-sandbox", "--disable-setuid-sandbox"]
        )

        page = await browser.new_page()

        # 1. 開 IR 頁
        await page.goto(IR_URL)
        await page.wait_for_timeout(5000)

        # 2. 抓所有 webcast event
        links = await page.query_selector_all("a")

        events = []
        for a in links:
            href = await a.get_attribute("href")
            if href and "zucast.com/webcast" in href:
                events.append(href)

        print("event數量:", len(events))

        if len(events) < 4:
            print("❌ 找不到第4個 event")
            await browser.close()
            return

        # 3. 第3個 event
        event_url = events[3]
        print("第4個event:", event_url)

        await page.goto(event_url)
        await page.wait_for_timeout(5000)

        # 4. 找表單（如果有）
        try:
            await page.wait_for_selector("input.first_name", timeout=5000)

            await page.fill("input.first_name", DATA["first_name"])
            await page.fill("input.last_name", DATA["last_name"])
            await page.fill("input.email", DATA["email"])
            await page.fill("input.telephone", DATA["telephone"])

            print("✔ 表單填完")

            await page.check("input.agreement")
            print("✔ agreement已勾選")

            await page.wait_for_timeout(2000)

            button = await page.query_selector("button.submit")
            disabled = await button.get_attribute("disabled")
            print("submit disabled:", disabled)


            # ⭐ submit（你補充的最後一步）
            await page.click("button.submit")
            print("✔ 已點 submit")

            # 等 redirect 進 webcast
            await page.wait_for_timeout(8000)

        except Exception as e:
            print("沒有表單或填表失敗:", e)

        # 5. 等 audio 出現
        try:
            await page.wait_for_selector(
                "audio",
                state="attached",
                timeout=10000)

            audio_src = await page.get_attribute("audio", "src")

            if audio_src and audio_src.startswith("//"):
                audio_src = "https:" + audio_src

            print("\n🎧 MP3 URL:")
            print(audio_src)

            # 6. 下載 mp3
            r = requests.get(audio_src)

            with open("cathay_webcast.mp3", "wb") as f:
                f.write(r.content)

            print("✔ MP3 已下載：cathay_webcast.mp3")

        except Exception as e:
            print("❌ 抓音檔失敗:", e)

        await browser.close()


# Colab / Jupyter 用這行
await main()
```

**whisper**

```python
import whisper

# 1. 載入模型
model = whisper.load_model("small")  # 可改 tiny / small / medium / large

print("模型載入完成")

# 2. 轉文字
audio_file = "cathay_webcast.mp3"

result = model.transcribe(
    audio_file,
    language="zh"   # 強制中文
)

# 3. 輸出結果
print("\n===== 轉錄結果 =====\n")
print(result["text"])
```

```python
with open("output.txt", "w", encoding="utf-8") as f:
    f.write(result["text"])

print("已輸出 output medium.txt")
```

**SenseVoice**

```python
from funasr import AutoModel

# 載入 SenseVoice Small
model = AutoModel(
    model="iic/SenseVoiceSmall",
    vad_model="fsmn-vad",
    vad_kwargs={"max_single_segment_time": 30000},
    trust_remote_code=True,
)

# 語音辨識
res = model.generate(
    input="cathay_webcast.mp3",
    cache={},
    language="auto",   # 自動辨識中英
    use_itn=True,
    batch_size_s=60,
)

print(res)
```

```python
text = res[0]["text"]

with open("cathay_transcript.txt", "w", encoding="utf-8") as f:
    f.write(text)

print("✔ 已輸出 cathay_transcript.txt")
```
